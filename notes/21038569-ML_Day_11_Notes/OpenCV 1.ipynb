```python
import cv2
```

```python
image = cv2.imread("img.jpg")
image

# imread --> Read the image
# Opencv reads every pixel from the image and converts it into numpy array
```

```python
from matplotlib import pyplot as plt
```

```python
plt.imshow(image)
plt.show()
```

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image_rgb)
plt.show()

# cvtColor --> Convert one color space into another
```

```python
image.shape  # (width, height, color channels)
```

```python
image.ndim
```

```python
cv2.imshow("Image", image)  # showing the image
cv2.waitKey(5000)  # image will be displayed for 5000 milliseconds (5 seconds)
cv2.destroyAllWindows()  # closes all opened windows


#waitKey --> Pause execution until you press key in keyboard
```

```python
cv2.imshow("Image", image)  # showing the image
cv2.waitKey(0)  # shows the image till we close the window
cv2.destroyAllWindows()  

# cv2.waitKey(0) --> keeps the image window open until the user presses any key on keyboard
```
