### Sentence Tokenization

```python
import nltk
```

```python
# nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize
```

```python
text = "Python is easy to learn. NLP is simple. ML is useful"
print(text)

sentences = sent_tokenize(text)
print(sentences)
```

```python
# word tokenization

from nltk.tokenize import word_tokenize

text = "Python is easy to learn. NLP is simple. ML is useful"
tokens = word_tokenize(text)
print(tokens)
```

## Stop words

```python
nltk.download("stopwords")
```

```python
from nltk.corpus import stopwords
```

```python
stop_words = set(stopwords.words("english"))
print(stop_words)
```

```python
text = "The student is studying Python in the classroom"

stop_words = set(stopwords.words("english"))

words = text.split()
print(words)

clean_text = " ".join(
    i for i in words
    if i.lower() not in stop_words
)

print(clean_text)
```

### Stemming

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
```

```python
words = ["Playing", "Played", "Plays", "Connection", "studies"]

for i in words:
    print(i, "-->", stemmer.stem(i))
```

### Lemmatization

```python
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()

words = ["Playing", "Played", "Plays", "Connection", "studies"]

for i in words:
    print(i, "-->", lemmatizer.lemmatize(i))
```

```python
lemmatizer = WordNetLemmatizer()

print("Playing -->", lemmatizer.lemmatize("playing", pos = 'v'))
print("Played -->", lemmatizer.lemmatize("played", pos = 'v'))
print("Studies -->", lemmatizer.lemmatize("studies", pos = 'v'))
print("Children -->", lemmatizer.lemmatize("children", pos = 'n'))
```
