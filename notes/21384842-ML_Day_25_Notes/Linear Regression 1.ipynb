```python
# y = 10x + 20
# m = 10, c = 20

from sklearn.linear_model import LinearRegression

X = [[1],[2],[3],[4],[5]]
y = [30,40,50,60,70]

model = LinearRegression()
model.fit(X,y)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)

print("Prediction:", model.predict([[6]]))
```
