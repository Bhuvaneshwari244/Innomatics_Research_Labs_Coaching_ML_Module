### Resize / Compressing Images

```python
boy_pic = cv2.imread("img.jpg")

cv2.imshow("Pic", boy_pic)  # showing the image
cv2.waitKey(0)  # shows the image till we close the window
cv2.destroyAllWindows() 
```

```python
boy_pic.shape
```

```python
# resize() --> Changes the dimensions of an image

resized = cv2.resize(boy_pic, (500, 500))
cv2.imshow("Pic", resized)  # showing the image
cv2.waitKey(0)  # shows the image till we close the window
cv2.destroyAllWindows() 

# width = 500, height = 500
```

```python
## Blurring the image

blurred = cv2.blur(boy_pic, ksize = (20,20))
cv2.imshow("Pic", blurred)  # showing the image
cv2.waitKey(0)  # shows the image till we close the window
cv2.destroyAllWindows() 
```

```python
# edge detection
# Canny() --> Detects object boundaries and higlights only the important edges

edge = cv2.Canny(boy_pic, 200, 200)
cv2.imshow("Pic", edge)  # showing the image
cv2.waitKey(0)  # shows the image till we close the window
cv2.destroyAllWindows() 
```

```python
# Image Rotation

rotated = cv2.rotate(boy_pic, cv2.ROTATE_90_CLOCKWISE)
cv2.imshow("Pic", rotated)  # showing the image
cv2.waitKey(0)  # shows the image till we close the window
cv2.destroyAllWindows() 
```

```python
# Flipping image

# cv2.flip(image, 1) --> Horizontal flip
# cv2.flip(image, 0) --> Vertical flip
# cv2.flip(image, -1) --> Rotate + Mirror

flipped = cv2.flip(boy_pic, 1)
cv2.imshow("Pic", flipped)  # showing the image
cv2.waitKey(0)  # shows the image till we close the window
cv2.destroyAllWindows() 
```

```python
## Threshold --> Converts and image into black and white

# if pixel > 127 --> becomes white (255)
# if pixel <= 127 --> becomes black

image = cv2.imread("img.jpg", 0)
resized = cv2.resize(image, (300, 300))
_,thresh = cv2.threshold(resized,127,255,cv2.THRESH_BINARY)

cv2.imshow("Pic", thresh)  # showing the image
cv2.waitKey(0)  # shows the image till we close the window
cv2.destroyAllWindows() 
```

```python
black = np.zeros((500, 800, 3))
cv2.line(black, (50,50),(300,300), (0, 255, 0), 5)
# (50,50) --> Starting point
# (300,300) --> Ending point
# (0, 255, 0) --> Green color
# 5 --> Line thickness

cv2.imshow("black_pic", black)  # showing the image
cv2.waitKey(0)  # shows the image till we close the window
cv2.destroyAllWindows()
```

```python
black = np.zeros((500, 800, 3))
cv2.rectangle(black, (100, 100), (300, 250), (255,0,0),3)

cv2.imshow("black_pic", black)  # showing the image
cv2.waitKey(0)  # shows the image till we close the window
cv2.destroyAllWindows()
```

```python
black = np.zeros((500, 800, 3))
cv2.rectangle(black, (100, 100), (300, 250), (255,0,0),-1)
# -1 --> it will fill rectangle

cv2.imshow("black_pic", black)  # showing the image
cv2.waitKey(0)  # shows the image till we close the window
cv2.destroyAllWindows()
```
