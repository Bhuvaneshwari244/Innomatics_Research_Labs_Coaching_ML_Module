# Punctuation Marks

```python
import string
```

```python
string.punctuation
```

```python
text = "The course is good, I learned a lot #@"

clean_text = text.translate(str.maketrans("", "", string.punctuation))
# str.maketrans("", "", string.punctuation) -->
# --> creates a tranclation table that tells python to remove all characters present in string.punctuation
print(clean_text)
```

```python
text = "The course is god!, I learned a lot #@"

pattern = r'[^a-zA-Z0-9]'  # !,#@
pattern
```

```python
import re

clean_text2 = re.sub(pattern, ' ', text)
clean_text2
```
