# Punctuation Marks

```python
import string
```

```python
string.punctuation
```

```python
text = "The course is good, I learned a lot #@"

clean_text = text.translate(str.maketrans("", "", string.punctuation))
# str.maketrans("", "", string.punctuation) -->
# --> creates a tranclation table that tells python to remove all characters present in string.punctuation
print(clean_text)
```

```python
text = "The course is god!, I learned a lot #@"

pattern = r'[^a-zA-Z0-9]'  # !,#@
pattern
```

```python
import re

clean_text2 = re.sub(pattern, ' ', text)
clean_text2
```

# Handling URL's

```python
text = "Visit https://www.innomatics.in for more information"

pattern = r'https?://\S+|www.\.\S+'     # http or https
# \S+ means one or more non-space characters

clean_text = re.sub(pattern, '', text)
print(clean_text)
```

# Handling HTML Tags

```python
text = "<h1> Welcome to our website.</h1>"

pattern = r'<[^>]+>'

clean_text = re.sub(pattern, '', text)
print(clean_text)
```

# Handling Digits / Numbers

```python
text = "My phone number is 9000000111"
pattern = r'\d+'

clean_text = re.sub(pattern,'', text)
print(clean_text)
```

# Sentence Tokenization

```python
import nltk
```

```python
nltk.download('punkt')
```

```python
from nltk.tokenize import sent_tokenize
```

```python
text = "Python is easy to learn. NLP is simple. ML is useful"
print(text)
```

```python
# divide text into multiple sentences
sentences = sent_tokenize(text)
print(sentences)
```

```python
# word tokenization

from nltk.tokenize import word_tokenize

text = "Python is easy to learn. NLP is simple. ML is useful"
tokens = word_tokenize(text)
print(tokens)
```

# Stop words

```python
nltk.download("stopwords")
```

```python
from nltk.corpus import stopwords
```

```python
stop_words = set(stopwords.words("english"))
print(stop_words)
```

```python
text = "The student is studying Python in the classroom"

stop_words = set(stopwords.words("english"))

words = text.split()
print(words)

clean_text = " ".join(
    i for i in words
    if i.lower() not in stop_words
)

print(clean_text)
```

# Stemming

```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()
```

```python
words = ["Playing", "Played", "Plays", "Connection", "studies"]

for i in words:
    print(i, "-->", stemmer.stem(i))
```

# Lemmatization

```python
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()

words = ["Playing", "Played", "Plays", "Connection", "studies"]

for i in words:
    print(i, "-->", lemmatizer.lemmatize(i))
```

```python
lemmatizer = WordNetLemmatizer()

words = ["Playing", "Played", "Plays", "Connection", "studies"]

print("Playing -->", lemmatizer.lemmatize("playing", pos = 'v'))
print("Played -->", lemmatizer.lemmatize("played", pos = 'v'))
print("Studies -->", lemmatizer.lemmatize("studies", pos = 'v'))
print("Children -->", lemmatizer.lemmatize("children", pos = 'n'))
```
