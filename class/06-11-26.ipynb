```python
from sklearn.metrics import roc_curve, roc_auc_score
from sklearn.preprocessing import label_binarize
import matplotlib.pyplot as plt


# Convert actual classes into One-vs-Rest format

y_test_binary = label_binarize(
    y_test,
    classes=[0, 1, 2]
)


# Get probability of each class

y_proba = knn.predict_proba(X_test_scaled)


# Create 3 subplots

fig, axes = plt.subplots(1, 3, figsize=(18, 5))


for i in range(3):

    # Calculate FPR and TPR

    fpr, tpr, thresholds = roc_curve(
        y_test_binary[:, i],
        y_proba[:, i]
    )

    # Calculate AUC

    auc = roc_auc_score(
        y_test_binary[:, i],
        y_proba[:, i]
    )

    # Plot ROC curve

    axes[i].plot(
        fpr,
        tpr,
        marker="o",
        label=f"AUC = {auc:.2f}"
    )

    # 45-degree random classifier line

    axes[i].plot(
        [0, 1],
        [0, 1],
        linestyle="--",
        label="Random Classifier"
    )

    # Labels and title

    axes[i].set_xlabel("False Positive Rate (FPR)")
    axes[i].set_ylabel("True Positive Rate (TPR)")

    axes[i].set_title(
        f"{iris.target_names[i]} vs Non-{iris.target_names[i]}"
    )

    axes[i].legend()
    axes[i].grid()


plt.tight_layout()
plt.show()
```
