```python
text = "This computr is very useful"

correct_text = str(TextBlob(text).correct())
print("Original text:", text)
print("Corrected text:", correct_text)
```

### Emoji Handling

```python
pip install emoji
```

```python
import emoji
```

```python
text = "The movie is good 😊"

converted_text = emoji.demojize(text)
print(converted_text)
```

```python
text = "The movie is good 😊"

converted_text = emoji.demojize(text, delimiters = ('', ''))
print(converted_text)
```

```python
text = "The movie is good 😊"

converted_text = emoji.replace_emoji(text, replace = '')
print(converted_text)
```
